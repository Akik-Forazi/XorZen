## 1. Configuration

**Edit this cell before running anything else.** All user-configurable settings are here.
Defaults are tuned for Kaggle 2x T4 (16 GB VRAM each) + 30 GB system RAM.

```python
# ╔══════════════════════════════════════════════════════════════════╗
# ║              XORZEN zero_50M TRAINING CONFIGURATION              ║
# ║         KAGGLE EDITION — T4 GPU (16 GB VRAM)                     ║
# ╚══════════════════════════════════════════════════════════════════╝

import os
from pathlib import Path

# === MODEL ===
# 'zero_50M' (50 M params, default) or 'zero_277M' (277 M params, flagship).
# Both fit on a single T4 with bf16 + gradient checkpointing.
MODEL_VARIANT = "zero_50M"
USE_TEST_MODE = False                  # True = fewer experts loaded (dev only)

# === DATASET ===
DATASET_NAME = "roneneldan/TinyStories"  # HuggingFace (public, CC-BY-4.0)
DATASET_SPLIT = "train"
DATASET_VAL_SPLIT = "validation"
DATASET_TEXT_COLUMN = "text"
# On Kaggle (30 GB RAM) we cap the train subset to keep tokenization memory in check.
# None = full dataset (~2.1 M stories, ~500 MB text + ~3 GB int32 token cache).
# Set an int (e.g. 500_000) for a smaller, faster run.
MAX_DATASET_SAMPLES = 500_000
DATASET_SEED = 42

# Optional: if you attached TinyStories as a Kaggle dataset, set the mount path here.
# When set, the notebook loads from this directory instead of downloading from HF.
# Example: KAGGLE_INPUT_TINYSTORIES_DIR = "/kaggle/input/tinystories"
KAGGLE_INPUT_TINYSTORIES_DIR = None

# === TOKENIZER ===
# Auto-select tokenizer based on MODEL_VARIANT to avoid vocab mismatch crashes.
#   zero_50M  (vocab 10000)  → xorzen_python_10k       (10K vocab, trained on English+Python+Math)
#   zero_277M (vocab 33898)  → xorzen_agi_tokenizer_65k (65K covers 33898; OOV clamped to UNK) — only if you switch MODEL_VARIANT
# Override by setting TOKENIZER_NAME manually below.
AUTO_SELECT_TOKENIZER = True
TOKENIZER_NAME = "xorzen_python_10k"  # default; auto-overridden below if AUTO_SELECT_TOKENIZER=True

# === TRAINING ===
# BATCH_SIZE is the GLOBAL batch yielded by the DataLoader. With USE_MULTI_GPU=True
# and 2x T4 visible, each GPU processes BATCH_SIZE / world_size samples per step.
# zero_50M with BATCH_SIZE=16 (8 per GPU) + gradient checkpointing fits comfortably
# in 14.6 GB T4. The local attention pathway materializes a full [B, H, S, S]
# attention matrix (8 * 1024 * 1024 * 4 bytes = 32 MB per layer per sample) so
# larger batches OOM without checkpointing.
# For zero_277M, drop to BATCH_SIZE=4.
BATCH_SIZE = 16
GRADIENT_ACCUMULATION_STEPS = 2
SEQUENCE_LENGTH = 1024
LEARNING_RATE = 3e-4
WEIGHT_DECAY = 0.01
WARMUP_STEPS = 100
# Kaggle GPU sessions have a ~9-hour wall-clock limit. Tune MAX_TRAINING_STEPS
# based on the per-step time reported by the benchmark cell.
# At ~1-2s/step (zero_50M, batch 16, 2x T4, grad-ckpt on), 5000 steps = ~1.5-3h.
MAX_TRAINING_STEPS = 5000
VALIDATION_INTERVAL = 500
CHECKPOINT_INTERVAL = 1000
GRADIENT_CLIPPING = 1.0
MIXED_PRECISION = "bf16"               # T4 supports bf16 (preferred). 'fp16' or 'fp32' also valid.
SEED = 42

# === MULTI-GPU ===
# Default: True (use both T4s). Now that the hass_block Python-loop bottleneck
# is fixed (commit 863fee9), DataParallel overhead is no longer dominant — the
# GPU actually does real work per replica instead of waiting on Python.
# Set to False only if you have a single T4 or want to debug single-GPU.
# Both xorzen bugs that previously blocked DataParallel are fixed upstream:
#   - routing.py device-mismatch in _update_expert_usage (commit cdc50a5)
#   - ModelOutput not iterable in gather() (commit b3a5178)
#   - ModelOutputDataParallel class (commit bafcd97)
#   - sliced_ffn dtype mismatch under autocast (commit a0ad931)
#   - hass_block per-batch Python loop (commit 863fee9)
USE_MULTI_GPU = False

# === DATA PIPELINE (CPU) ===
# Kaggle gives 4 CPU cores. We use 6 workers (slight oversubscription helps
# hide data-loading latency from the GPU) + prefetch_factor=2 so 12 batches are
# always ready in the queue. This keeps the CPU pipeline ahead of the GPU so
# the GPU never waits for data.
NUM_WORKERS = 6
PREFETCH_FACTOR = 2

# === GRADIENT CHECKPOINTING ===
# OFF. Model only uses 3.5 GB VRAM — fits in 14.6 GB T4 without checkpointing.
# With seq=512 + batch=32 + no checkpointing, peak VRAM ~6-8 GB. Fine.
# Checkpointing adds ~34% overhead (recomputes forward during backward).
# Turn ON only for zero_277M or if you push BATCH_SIZE > 48.
FORCE_GRADIENT_CHECKPOINTING = False

# === TORCH.COMPILE ===
# JIT-compile the model with PyTorch 2.x. Fuses Python loops in MoE dispatch,
# SlicedFFN, and SSM scan into efficient C++ kernels. 3-5x speedup expected.
# First 2-3 forward passes are slow (compilation), then fast.
USE_TORCH_COMPILE = True

# === KAGGLE PATHS ===
# /kaggle/working  -> persisted output (downloadable from the Kaggle UI)
# /kaggle/input    -> read-only attached datasets
# /kaggle/temp     -> scratch (wiped at session end)
KAGGLE_WORKING_DIR = "/kaggle/working"
KAGGLE_INPUT_DIR = "/kaggle/input"
KAGGLE_TEMP_DIR = "/kaggle/temp"
OUTPUT_BASE_DIR = f"XORZEN/{MODEL_VARIANT}"   # subfolder under /kaggle/working

# === GENERATION ===
GENERATION_MAX_NEW_TOKENS = 128
GENERATION_TEMPERATURE = 0.8
GENERATION_TOP_K = 50
GENERATION_TOP_P = 0.9

# === INSTALLATION ===
# Set XORZEN_WHEEL_PATH to a local .whl path (e.g. attached as a Kaggle dataset)
# to install offline. Leave None to install from GitHub (requires Internet=On).
XORZEN_WHEEL_PATH = None
XORZEN_GITHUB_URL = "git+https://github.com/Akik-Forazi/XorZen.git"

# === AUTO-SELECT TOKENIZER (do not edit) ===
# Resolve tokenizer based on MODEL_VARIANT to avoid vocab mismatch crashes.
if AUTO_SELECT_TOKENIZER:
    _tokenizer_for_variant = {
        "zero_50M":  "xorzen_python_10k",            # 10K vocab, trained on English+Python+Math
        "zero_277M": "xorzen_agi_tokenizer_65k",  # 65K vocab covers zero_277M (33898) — only if you switch
    }
    _chosen = _tokenizer_for_variant.get(MODEL_VARIANT, TOKENIZER_NAME)
    if _chosen != TOKENIZER_NAME:
        print(f"  Auto-selected tokenizer '{_chosen}' for MODEL_VARIANT='{MODEL_VARIANT}'")
        TOKENIZER_NAME = _chosen

print("Configuration loaded (Kaggle 2x T4 edition)")
print(f"  Model: {MODEL_VARIANT}")
print(f"  Tokenizer: {TOKENIZER_NAME}")
print(f"  Effective batch size: {BATCH_SIZE * GRADIENT_ACCUMULATION_STEPS}")
print(f"  Max steps: {MAX_TRAINING_STEPS}")
print(f"  Mixed precision: {MIXED_PRECISION}")
print(f"  Multi-GPU: {USE_MULTI_GPU}")
print(f"  Grad checkpointing: {FORCE_GRADIENT_CHECKPOINTING}")
print(f"  torch.compile: {USE_TORCH_COMPILE}")
print(f"  Num workers (CPU): {NUM_WORKERS}, prefetch: {PREFETCH_FACTOR}")
print(f"  Max dataset samples: {MAX_DATASET_SAMPLES}")
```

## 3. Install XORZEN

Install the XORZEN framework from GitHub (requires Kaggle Internet=On) or from a
pre-attached local wheel (offline mode).

```python
import subprocess, sys, os
from pathlib import Path

def install_xorzen():
    print("Installing XORZEN dependencies...")
    # Install only the deps xorzen needs BESIDES torch/torchvision — those
    # are already provided by Kaggle with CUDA support, and pip reinstalling
    # them would replace the CUDA build with a CPU-only build, breaking GPU.
    core_deps = ["tokenizers>=0.21", "sentencepiece", "einops", "pydantic>=2.0",
                 "psutil", "scipy", "datasets", "matplotlib", "numpy",
                 "cryptography"]
    for dep in core_deps:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", dep])

    # Tell HuggingFace to cache inside /kaggle/working so caches persist in output
    hf_cache = Path(KAGGLE_WORKING_DIR) / "hf_cache"
    hf_cache.mkdir(parents=True, exist_ok=True)
    os.environ.setdefault("HF_HOME", str(hf_cache))
    os.environ.setdefault("HF_DATASETS_CACHE", str(hf_cache / "datasets"))
    os.environ.setdefault("TRANSFORMERS_CACHE", str(hf_cache / "transformers"))

    if XORZEN_WHEEL_PATH and os.path.exists(XORZEN_WHEEL_PATH):
        print(f"  Installing from local wheel: {XORZEN_WHEEL_PATH}")
        # --no-deps to avoid pip replacing Kaggle's CUDA torch
        subprocess.check_call([sys.executable, "-m", "pip", "install", "--no-deps", "-q", XORZEN_WHEEL_PATH])
    else:
        # Try GitHub install (requires Internet=On in Kaggle settings)
        # CRITICAL: use --no-deps so pip doesn't reinstall torch as CPU-only.
        # Kaggle ships a CUDA-enabled torch; pip install git+... would pull
        # plain 'torch' from PyPI which is CPU-only, breaking torch.cuda.
        print(f"  Installing from GitHub: {XORZEN_GITHUB_URL}")
        local_clone = Path(KAGGLE_WORKING_DIR) / "XorZen"
        try:
            subprocess.check_call([sys.executable, "-m", "pip", "install", "--no-deps", "-q", XORZEN_GITHUB_URL])
        except subprocess.CalledProcessError as e:
            print(f"  GitHub install failed ({e}). Trying local clone fallback...")
            if local_clone.exists():
                print(f"  Using existing clone at {local_clone}")
                subprocess.check_call([sys.executable, "-m", "pip", "install", "--no-deps", "-q", "-e", str(local_clone)])
            else:
                print("  Cloning XorZen repo into /kaggle/working/XorZen...")
                subprocess.check_call(["git", "clone", "--depth", "1",
                                       "https://github.com/Akik-Forazi/XorZen.git",
                                       str(local_clone)])
                subprocess.check_call([sys.executable, "-m", "pip", "install", "--no-deps", "-q", "-e", str(local_clone)])
    print("XORZEN installed")
    # Verify torch.cuda survived the install
    import torch
    print(f"  torch version: {torch.__version__}")
    print(f"  torch.cuda.is_available(): {torch.cuda.is_available()}")
    if not torch.cuda.is_available():
        print("  WARNING: torch.cuda is NOT available! The xorzen install may have")
        print("           replaced Kaggle's CUDA torch with a CPU-only build.")
        print("           Fix: restart kernel, re-run this cell, check that")
        print("           subprocess uses --no-deps (it now does).")

install_xorzen()
```

## 7. Prepare Dataset

Tokenize text and pack into fixed-length sequences. Token caches are written to
`/kaggle/working/.../tokenizer_cache/` so they persist across sessions.

**OOV clamping**: if the tokenizer's vocab is larger than the model's vocab
(e.g. 65K tokenizer + 33898-vocab 277M model — only relevant if you switch MODEL_VARIANT), tokens whose IDs exceed the
model's vocab are replaced with the UNK token to avoid `IndexError` in
`nn.Embedding`. The cache filename includes the clamp threshold so caches
aren't reused across model variants.

```python
import torch
from torch.utils.data import Dataset, DataLoader
import numpy as np
from pathlib import Path
import time
import pickle

class XORZENTextDataset(Dataset):
    """Tokenized text dataset for XORZEN language model training."""

    def __init__(self, texts, tokenizer, sequence_length, cache_path=None, max_token_id=None, unk_token_id=1):
        self.sequence_length = sequence_length
        self.max_token_id = max_token_id  # if set, OOV tokens (id > max_token_id) → unk_token_id
        self.unk_token_id = unk_token_id

        # Cache filename includes max_token_id so a cache saved with clamping isn't reused without it
        effective_cache = cache_path
        if cache_path and max_token_id is not None:
            p = Path(cache_path)
            effective_cache = p.with_name(p.stem + f"_clamp{max_token_id}" + p.suffix)

        if effective_cache and Path(effective_cache).exists():
            print(f"  Loading cached tokens from {effective_cache}")
            t0 = time.time()
            with open(effective_cache, "rb") as f:
                self.all_tokens = pickle.load(f)
            print(f"  Cache loaded in {time.time()-t0:.1f}s ({len(self.all_tokens):,} tokens)")
        else:
            print(f"  Tokenizing {len(texts):,} texts...")
            t0 = time.time()
            all_tokens = []
            for i, text in enumerate(texts):
                if text and len(text.strip()) > 0:
                    token_ids = tokenizer.encode(text)
                    all_tokens.extend(token_ids)
                if (i + 1) % 50000 == 0:
                    print(f"    {i+1:,}/{len(texts):,} texts ({len(all_tokens):,} tokens)")
            self.all_tokens = np.array(all_tokens, dtype=np.int32)
            elapsed = time.time() - t0
            print(f"  Tokenization done: {len(self.all_tokens):,} tokens in {elapsed:.1f}s")
            del all_tokens

            # Clamp OOV tokens to UNK to avoid IndexError in nn.Embedding
            # (tokenizer vocab may exceed model vocab; only matters if you switch to zero_277M with the 65K tokenizer)
            if max_token_id is not None:
                n_oov = int((self.all_tokens > max_token_id).sum())
                if n_oov > 0:
                    pct = 100.0 * n_oov / len(self.all_tokens)
                    print(f"  Clamping {n_oov:,} OOV tokens (>{max_token_id}) → UNK({unk_token_id}) "
                          f"[{pct:.3f}% of tokens]")
                    self.all_tokens = np.where(self.all_tokens > max_token_id,
                                                np.int32(unk_token_id), self.all_tokens)

            if effective_cache:
                Path(effective_cache).parent.mkdir(parents=True, exist_ok=True)
                with open(effective_cache, "wb") as f:
                    pickle.dump(self.all_tokens, f)
                print(f"  Cache saved to {effective_cache}")

        n_tokens = len(self.all_tokens)
        self.n_samples = max(0, (n_tokens - 1) // sequence_length)
        print(f"  Packed into {self.n_samples:,} sequences of length {sequence_length}")

    def __len__(self):
        return self.n_samples

    def __getitem__(self, idx):
        start = idx * self.sequence_length
        end = start + self.sequence_length + 1
        chunk = self.all_tokens[start:end]
        input_ids = torch.tensor(chunk[:-1], dtype=torch.long)
        labels = torch.tensor(chunk[1:], dtype=torch.long)
        return {"input_ids": input_ids, "labels": labels}


# Build datasets
print("=" * 70)
print("Dataset Preparation")
print("=" * 70)

cache_dir = Path(str(CHECKPOINT_DIR)) / "tokenizer_cache"
cache_dir.mkdir(parents=True, exist_ok=True)

# Determine the model's vocab_size so we can clamp OOV tokens to UNK.
# `config` was set in the tokenizer cell (cell 13).
MODEL_VOCAB_SIZE = config.vocab_size
MAX_TOKEN_ID = MODEL_VOCAB_SIZE - 1
UNK_TOKEN_ID = 1
print(f"  Model vocab_size: {MODEL_VOCAB_SIZE}  → max allowed token id: {MAX_TOKEN_ID}")
print(f"  Tokenizer vocab:  {tokenizer.get_vocab_size()}")
if tokenizer.get_vocab_size() > MODEL_VOCAB_SIZE:
    print(f"  NOTE: tokenizer > model vocab — OOV tokens will be clamped to UNK({UNK_TOKEN_ID})")

print("\nExtracting training texts...")
if MAX_DATASET_SAMPLES and MAX_DATASET_SAMPLES < len(train_dataset_raw):
    train_texts = train_dataset_raw.select(range(min(MAX_DATASET_SAMPLES, len(train_dataset_raw))))
else:
    train_texts = train_dataset_raw
train_text_list = [ex[DATASET_TEXT_COLUMN] for ex in train_texts]
print(f"  Training examples: {len(train_text_list):,}")

print("\nPreparing training dataset:")
train_cache = cache_dir / f"train_{TOKENIZER_NAME}_{SEQUENCE_LENGTH}_{len(train_text_list)}.pkl"
train_dataset = XORZENTextDataset(
    train_text_list, tokenizer, SEQUENCE_LENGTH, cache_path=str(train_cache),
    max_token_id=MAX_TOKEN_ID, unk_token_id=UNK_TOKEN_ID,
)
del train_text_list  # free RAM

val_dataset = None
if val_dataset_raw:
    print("\nPreparing validation dataset:")
    # Cap val set to keep RAM in check on Kaggle
    val_cap = 50_000
    if len(val_dataset_raw) > val_cap:
        print(f"  Capping validation set to {val_cap:,} samples (RAM-safe).")
        val_dataset_raw = val_dataset_raw.select(range(val_cap))
    val_text_list = [ex[DATASET_TEXT_COLUMN] for ex in val_dataset_raw]
    print(f"  Validation examples: {len(val_text_list):,}")
    val_cache = cache_dir / f"val_{TOKENIZER_NAME}_{SEQUENCE_LENGTH}_{len(val_text_list)}.pkl"
    val_dataset = XORZENTextDataset(
        val_text_list, tokenizer, SEQUENCE_LENGTH, cache_path=str(val_cache),
        max_token_id=MAX_TOKEN_ID, unk_token_id=UNK_TOKEN_ID,
    )
    del val_text_list

print(f"\n--- Dataset Summary ---")
print(f"  Sequence length: {SEQUENCE_LENGTH}")
print(f"  Training samples: {len(train_dataset):,}")
print(f"  Training tokens:  {len(train_dataset) * SEQUENCE_LENGTH:,}")
if val_dataset:
    print(f"  Val samples:      {len(val_dataset):,}")
    print(f"  Val tokens:       {len(val_dataset) * SEQUENCE_LENGTH:,}")

sample = train_dataset[0]
print(f"\n  Sample input_ids shape: {sample['input_ids'].shape}")
print(f"  Sample input_ids[:20]:  {sample['input_ids'][:20].tolist()}")
decoded_sample = tokenizer.decode(sample['input_ids'][:100].tolist())
print(f"  Decoded[:100]: '{decoded_sample[:100]}...'")

print(f"\nOK: Datasets ready")
print("=" * 70)
```

## 8. Initialize the model

Instantiate the model on cuda:0, then wrap with `ModelOutputDataParallel`
(from `xorzen.model.base`, commit `bafcd97`) when 2+ T4s are visible and
`USE_MULTI_GPU=True`. This subclass averages scalar loss tensors across
replicas so `outputs.loss.item()` works — stock `nn.DataParallel` would
concatenate them into a 1-D tensor and crash `.item()`.

All xorzen bugs that previously blocked multi-GPU training are now fixed
upstream:
- `routing.py` device-mismatch in `_update_expert_usage` (commit `cdc50a5`)
- `ModelOutput` not iterable in `gather()` (commit `b3a5178`)
- `ModelOutputDataParallel` class added (commit `bafcd97`)
- `sliced_ffn` dtype mismatch under autocast (commit `a0ad931`)

If you have an older xorzen install, run:
`pip install --upgrade --force-reinstall git+https://github.com/Akik-Forazi/XorZen.git`

```python
# ╔══════════════════════════════════════════════════════════════════╗
# ║                  XORZEN MODEL INITIALIZATION                    ║
# ║   KAGGLE T4 — multi-GPU via ModelOutputDataParallel              ║
# ╚══════════════════════════════════════════════════════════════════╝

import torch
import torch.nn as nn
import time
import gc
import importlib
from xorzen.model.base import ModelOutputDataParallel

print("=" * 70)
print(f"XORZEN {MODEL_VARIANT} Model Initialization")
print("=" * 70)

# Note: All xorzen bugs that previously blocked multi-GPU training are now
# fixed in the upstream repo (https://github.com/Akik-Forazi/XorZen):
#   - commit cdc50a5: routing.py device-mismatch in _update_expert_usage
#   - commit b3a5178: ModelOutput __iter__ + custom __init__ for gather()
#   - commit bafcd97: ModelOutputDataParallel class (averages scalar losses)
#   - commit a0ad931: sliced_ffn dtype mismatch under autocast
# If you have an older xorzen install, run:
#   pip install --upgrade --force-reinstall git+https://github.com/Akik-Forazi/XorZen.git

torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Resolve device: with DataParallel we still move the model to cuda:0 first.
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"\nPrimary device: {device}")
print(f"Visible GPUs:   {torch.cuda.device_count()}")
print(f"USE_MULTI_GPU:  {USE_MULTI_GPU_EFFECTIVE}")

# Instantiate the ACTUAL model based on MODEL_VARIANT
print(f"\nInstantiating xorzen.{MODEL_VARIANT}(test_mode={USE_TEST_MODE})...")
t0 = time.time()
try:
    model_module_name = "xorzen.models.zero"
    model_class = getattr(importlib.import_module(model_module_name), MODEL_VARIANT)
    model = model_class(test_mode=USE_TEST_MODE)
except (AttributeError, ImportError) as e:
    print(f"ERROR: Could not instantiate '{MODEL_VARIANT}' from '{model_module_name}'. Defaulting to zero_50M. Error: {e}")
    from xorzen.models.zero import zero_50M
    model = zero_50M(test_mode=USE_TEST_MODE)

init_time = time.time() - t0
print(f"Model created in {init_time:.1f}s")

# Force gradient checkpointing on or off based on the config flag.
# The model's config defaults to gradient_checkpointing=True, so we must
# EXPLICITLY set it to False when FORCE_GRADIENT_CHECKPOINTING=False —
# otherwise the model keeps checkpointing on even when the user disables it.
if hasattr(model, 'gradient_checkpointing'):
    model.gradient_checkpointing = FORCE_GRADIENT_CHECKPOINTING
    print(f"  Set gradient_checkpointing={FORCE_GRADIENT_CHECKPOINTING}")
elif hasattr(model, 'config') and hasattr(model.config, 'gradient_checkpointing'):
    model.config.gradient_checkpointing = FORCE_GRADIENT_CHECKPOINTING
    if hasattr(model, 'gradient_checkpointing'):
        model.gradient_checkpointing = FORCE_GRADIENT_CHECKPOINTING
    print(f"  Set gradient_checkpointing={FORCE_GRADIENT_CHECKPOINTING} via config")

print(f"Moving to {device}...")
model.to(device)

# Wrap with ModelOutputDataParallel for multi-GPU if 2+ GPUs are visible.
# This subclass (from xorzen.model.base, commit bafcd97) averages scalar
# tensor fields (loss, routing_loss, etc.) across replicas so outputs.loss
# stays a 0-D scalar and .item() works. Stock nn.DataParallel would concat
# them into a 1-D tensor and crash .item().
if USE_MULTI_GPU_EFFECTIVE and torch.cuda.device_count() >= 2:
    device_ids = list(range(torch.cuda.device_count()))
    print(f"\nWrapping model with ModelOutputDataParallel(device_ids={device_ids})...")
    model = ModelOutputDataParallel(model, device_ids=device_ids)
    IS_DATA_PARALLEL = True
else:
    IS_DATA_PARALLEL = False
    if USE_MULTI_GPU and not USE_MULTI_GPU_EFFECTIVE:
        print("\n  (DataParallel disabled: only 1 GPU visible)")

# Helper accessor for the underlying model regardless of DP wrapping
def get_raw_model(m):
    return m.module if isinstance(m, ModelOutputDataParallel) else m

raw_model = get_raw_model(model)

# =====================================================================
# torch.compile() — JIT-compile the model to fuse Python loops into C++.
# This is the REAL speedup for xorzen. The architecture has Python-level
# loops in MoE dispatch, SlicedFFN width grouping, and SSM scan that
# torch.compile fuses into efficient C++ kernels.
#
# Mode='reduce-overhead' also uses CUDA graphs for even less Python overhead.
# Expected: 3-5x speedup on top of the architecture fixes already applied.
#
# If torch.compile fails (some ops don't support it), the model falls back
# to eager mode automatically. The first 2-3 forward passes will be slow
# (compilation), then subsequent passes are fast.
# =====================================================================
USE_TORCH_COMPILE = True
if USE_TORCH_COMPILE and hasattr(torch, 'compile'):
    try:
        print("\nCompiling model with torch.compile(mode='reduce-overhead')...")
        print("  (first 2-3 forward passes will be slow during compilation)")
        raw_model = torch.compile(raw_model, mode='reduce-overhead')
        model = raw_model  # no DP wrapping, so model == raw_model
        print("  torch.compile() successful — subsequent forward passes will be fast")
    except Exception as e:
        print(f"  torch.compile() failed (falling back to eager mode): {e}")
        print(f"  The model will work but will be slower than compiled mode.")
else:
    print("\n  torch.compile() not available (PyTorch < 2.0) — using eager mode")

total_params = raw_model.count_parameters()
trainable_params = raw_model.count_parameters(only_trainable=True)
print(f"\n--- Model Statistics ---")
print(f"  Total parameters:     {total_params:,}")
print(f"  Trainable parameters: {trainable_params:,}")
print(f"  Frozen parameters:    {total_params - trainable_params:,}")
footprint = raw_model.get_memory_footprint()
print(f"  Memory footprint:     {footprint['total_size_mb']:.1f} MB")

cfg = raw_model.config
print(f"\n--- Architecture ---")
for attr in ["model_name", "vocab_size", "hidden_size", "num_layers",
             "num_attention_heads", "context_length", "expert_count",
             "top_k_experts", "width_choices", "max_depth", "min_depth",
             "cot_dim", "cot_components", "ssm_state_dim",
             "local_window_size", "low_rank_dim", "tie_word_embeddings",
             "gradient_checkpointing"]:
    print(f"  {attr}: {getattr(cfg, attr, 'N/A')}")
if hasattr(raw_model, '_cot_enabled'):
    print(f"  cot_enabled: {raw_model._cot_enabled}")

# Forward pass test
print(f"\n--- Forward Pass Test ---")
batch_size = 2
seq_len = min(SEQUENCE_LENGTH, 128)
test_ids = torch.randint(0, cfg.vocab_size, (batch_size, seq_len), device=device)
test_lbl = torch.randint(0, cfg.vocab_size, (batch_size, seq_len), device=device)

t0 = time.time()
raw_model.eval()
with torch.no_grad():
    outputs = model(input_ids=test_ids, labels=test_lbl, return_dict=True)
fwd_time = time.time() - t0
print(f"  Forward time: {fwd_time:.3f}s")
print(f"  Logits shape: {outputs.logits.shape}")
print(f"  Loss: {outputs.loss.item():.4f}")
if hasattr(outputs, 'routing_loss') and outputs.routing_loss is not None:
    print(f"  Routing loss: {outputs.routing_loss.item():.6f}")
if hasattr(outputs, 'load_balance_loss') and outputs.load_balance_loss is not None:
    print(f"  Load balance loss: {outputs.load_balance_loss.item():.6f}")

# Backward pass test
print(f"\n--- Backward Pass Test ---")
raw_model.train()
ids2 = torch.randint(0, cfg.vocab_size, (batch_size, seq_len), device=device)
lbl2 = torch.randint(0, cfg.vocab_size, (batch_size, seq_len), device=device)
t0 = time.time()
out2 = model(input_ids=ids2, labels=lbl2, return_dict=True)
loss = out2.loss
loss.backward()
bwd_time = time.time() - t0
print(f"  Backward time: {bwd_time:.3f}s")
print(f"  Loss: {loss.item():.4f}")

grads_exist = 0
grads_nan = 0
grads_inf = 0
missing_grad = []
for name, param in raw_model.named_parameters():
    if param.grad is not None:
        grads_exist += 1
        if torch.isnan(param.grad).any(): grads_nan += 1
        if torch.isinf(param.grad).any(): grads_inf += 1
    elif param.requires_grad:
        missing_grad.append(name)

print(f"  Params with gradients: {grads_exist}")
if grads_nan: print(f"  WARNING: NaN gradients: {grads_nan}")
if grads_inf: print(f"  WARNING: Inf gradients: {grads_inf}")
if missing_grad: print(f"  WARNING: Missing gradients for: {missing_grad[:5]}")

del test_ids, test_lbl, outputs, ids2, lbl2, out2, loss
raw_model.zero_grad()
if torch.cuda.is_available(): torch.cuda.empty_cache()
gc.collect()

FORWARD_OK = True
BACKWARD_OK = grads_nan == 0 and grads_inf == 0
print(f"\n{'OK' if FORWARD_OK else 'FAIL'}: Forward pass")
print(f"{'OK' if BACKWARD_OK else 'WARN'}: Backward pass")
print("=" * 70)
```

## 9. Pre-flight Tests

Small pre-training sanity batch to verify end-to-end loss + backward works in mixed precision.

```python
import torch
from torch.optim import AdamW
from torch.utils.data import DataLoader
import gc

print("=" * 70)
print("Pre-flight Training Sanity Test")
print("=" * 70)

raw_model.train()
optimizer = AdamW(
    [p for p in model.parameters() if p.requires_grad],
    lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY, betas=(0.9, 0.95),
)

# Setup mixed precision (PyTorch 2.x API)
scaler = None
autocast_dtype = None
if MIXED_PRECISION == "fp16" and device.type == "cuda":
    scaler = torch.amp.GradScaler('cuda')
    autocast_dtype = torch.float16
elif MIXED_PRECISION == "bf16" and device.type == "cuda":
    autocast_dtype = torch.bfloat16
print(f"Mixed precision: {MIXED_PRECISION}")

sanity_samples = 8
sanity_ds = torch.utils.data.Subset(train_dataset, range(min(sanity_samples, len(train_dataset))))
sanity_loader = DataLoader(sanity_ds, batch_size=2, shuffle=False)

losses = []
for step, batch in enumerate(sanity_loader):
    input_ids = batch["input_ids"].to(device)
    labels = batch["labels"].to(device)
    optimizer.zero_grad()

    if autocast_dtype is not None:
        with torch.autocast(device_type=device.type, dtype=autocast_dtype):
            outputs = model(input_ids=input_ids, labels=labels, return_dict=True)
            loss = outputs.loss
    else:
        outputs = model(input_ids=input_ids, labels=labels, return_dict=True)
        loss = outputs.loss

    if torch.isnan(loss) or torch.isinf(loss):
        print(f"  Step {step}: LOSS IS {'NaN' if torch.isnan(loss) else 'Inf'}!")
        break

    if scaler is not None:
        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), GRADIENT_CLIPPING)
        scaler.step(optimizer)
        scaler.update()
    else:
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), GRADIENT_CLIPPING)
        optimizer.step()

    losses.append(loss.item())
    print(f"  Step {step}: loss = {loss.item():.4f}")
    del input_ids, labels, outputs, loss

PREFLIGHT_OK = len(losses) == len(sanity_loader) and all(l == l for l in losses)
print(f"\n--- Sanity Results ---")
print(f"  Steps: {len(losses)}")
if losses:
    print(f"  Final loss: {losses[-1]:.4f}")
    print(f"  Range: [{min(losses):.4f}, {max(losses):.4f}]")
    print(f"  Decreasing: {losses[-1] < losses[0]}")
print(f"\n{'OK' if PREFLIGHT_OK else 'FAIL'}: Pre-flight test")
print("=" * 70)
```

```python
import torch
import time
from torch.utils.data import DataLoader

print("=" * 70)
print("Performance Benchmark")
print("=" * 70)

raw_model.train()
bench_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
bench_iter = iter(bench_loader)

# Warmup
print(f"\nWarming up (3 steps)...")
for _ in range(3):
    try:
        batch = next(bench_iter)
    except StopIteration:
        bench_iter = iter(bench_loader)
        batch = next(bench_iter)
    input_ids = batch["input_ids"].to(device)
    labels = batch["labels"].to(device)
    optimizer.zero_grad()
    if autocast_dtype is not None:
        with torch.autocast(device_type=device.type, dtype=autocast_dtype):
            out = model(input_ids=input_ids, labels=labels, return_dict=True)
            l = out.loss / GRADIENT_ACCUMULATION_STEPS
    else:
        out = model(input_ids=input_ids, labels=labels, return_dict=True)
        l = out.loss / GRADIENT_ACCUMULATION_STEPS
    if scaler: scaler.scale(l).backward()
    else: l.backward()
    if scaler: scaler.unscale_(optimizer)
    torch.nn.utils.clip_grad_norm_(model.parameters(), GRADIENT_CLIPPING)
    if scaler: scaler.step(optimizer); scaler.update()
    else: optimizer.step()
    del input_ids, labels, out, l

if torch.cuda.is_available(): torch.cuda.empty_cache()

# Measure
BENCH_STEPS = 10
print(f"\nBenchmarking ({BENCH_STEPS} steps, batch_size={BATCH_SIZE})...")
step_times = []
gpu_alloc = []

for _ in range(BENCH_STEPS):
    try:
        batch = next(bench_iter)
    except StopIteration:
        bench_iter = iter(bench_loader)
        batch = next(bench_iter)
    input_ids = batch["input_ids"].to(device)
    labels = batch["labels"].to(device)
    optimizer.zero_grad()
    t_start = time.time()
    if autocast_dtype is not None:
        with torch.autocast(device_type=device.type, dtype=autocast_dtype):
            out = model(input_ids=input_ids, labels=labels, return_dict=True)
            l = out.loss / GRADIENT_ACCUMULATION_STEPS
    else:
        out = model(input_ids=input_ids, labels=labels, return_dict=True)
        l = out.loss / GRADIENT_ACCUMULATION_STEPS
    if scaler: scaler.scale(l).backward()
    else: l.backward()
    if scaler: scaler.unscale_(optimizer)
    torch.nn.utils.clip_grad_norm_(model.parameters(), GRADIENT_CLIPPING)
    if scaler: scaler.step(optimizer); scaler.update()
    else: optimizer.step()
    if torch.cuda.is_available(): torch.cuda.synchronize()
    t_end = time.time()
    step_times.append(t_end - t_start)
    if torch.cuda.is_available():
        gpu_alloc.append(torch.cuda.memory_allocated() / (1024**3))
    del input_ids, labels, out, l

avg_step = sum(step_times) / len(step_times)
avg_tps = (BATCH_SIZE * SEQUENCE_LENGTH) / avg_step
peak_gpu = max(gpu_alloc) if gpu_alloc else 0

print(f"\n--- Benchmark Results ---")
print(f"  Avg step time:   {avg_step:.3f}s")
print(f"  Throughput:      {avg_tps:,.0f} tokens/sec (single micro-batch)")
print(f"  Peak GPU alloc:  {peak_gpu:.2f} GB (on cuda:0)")
if torch.cuda.is_available():
    total_vram = torch.cuda.get_device_properties(0).total_memory / (1024**3)
    print(f"  GPU0 utilization: {peak_gpu/total_vram*100:.1f}%")
    print(f"  GPUs visible:    {torch.cuda.device_count()}")

print(f"\n--- Time Estimates (per optimizer step ~ {avg_step * GRADIENT_ACCUMULATION_STEPS:.2f}s) ---")
for n in [1000, 5000, 10000, 50000]:
    hrs = n * avg_step * GRADIENT_ACCUMULATION_STEPS / 3600
    fits = "OK" if hrs < 9 else "EXCEEDS Kaggle 9h limit"
    print(f"  {n:,} steps: ~{hrs:.1f} hours [{fits}]")
print(f"  Configured run ({MAX_TRAINING_STEPS:,} steps): "
      f"~{MAX_TRAINING_STEPS * avg_step * GRADIENT_ACCUMULATION_STEPS / 3600:.1f} hours")

BENCHMARK_OK = True
benchmark_results = {"avg_step_time": avg_step, "avg_tokens_per_sec": avg_tps,
                     "peak_gpu_gb": peak_gpu, "num_gpus": torch.cuda.device_count() if torch.cuda.is_available() else 0}
print(f"\nOK: Benchmark complete")
print("=" * 70)
```

## 11. Train

Real training loop with mixed precision, gradient accumulation, cosine LR schedule,
periodic validation, and resumable checkpoints written to `/kaggle/working/`.

```python
import torch
from torch.optim import AdamW
from torch.optim.lr_scheduler import LambdaLR
from torch.utils.data import DataLoader
import time, json, math, gc
from collections import defaultdict
from pathlib import Path

print("=" * 70)
print(f"XORZEN {MODEL_VARIANT} Training (Kaggle T4)")
print("=" * 70)

# DataLoaders — CPU pipeline feeds both GPUs.
# NUM_WORKERS=6 with PREFETCH_FACTOR=2 keeps 12 batches queued up so the
# GPUs never wait for data. pin_memory=True enables fast async H2D copies.
_num_workers = NUM_WORKERS if device.type == "cuda" else 0
_prefetch = PREFETCH_FACTOR if _num_workers > 0 else None
_persistent = (_num_workers > 0)
train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=_num_workers, pin_memory=(device.type == "cuda"),
    drop_last=True, persistent_workers=_persistent,
    prefetch_factor=_prefetch,
)
val_loader = None
if val_dataset:
    val_loader = DataLoader(
        val_dataset, batch_size=BATCH_SIZE, shuffle=False,
        num_workers=_num_workers, pin_memory=(device.type == "cuda"),
        drop_last=False, persistent_workers=_persistent,
        prefetch_factor=_prefetch,
    )

# Fresh optimizer over all trainable params (works for both DP and non-DP)
raw_model.train()
optimizer = AdamW(
    [p for p in model.parameters() if p.requires_grad],
    lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY, betas=(0.9, 0.95), eps=1e-8,
)

# Cosine LR with warmup
def lr_lambda(step):
    if step < WARMUP_STEPS:
        return step / max(1, WARMUP_STEPS)
    progress = (step - WARMUP_STEPS) / max(1, MAX_TRAINING_STEPS - WARMUP_STEPS)
    return 0.5 * (1.0 + math.cos(math.pi * min(progress, 1.0)))

scheduler = LambdaLR(optimizer, lr_lambda)

# Training state
global_step = 0
best_val_loss = float('inf')
training_metrics = defaultdict(list)
start_time = time.time()
OOM_OCCURRED = False

# Resume from checkpoint
LATEST_CHECKPOINT = None
if CHECKPOINT_DIR:
    ckpts = sorted(Path(CHECKPOINT_DIR).glob("training_checkpoint_step_*.pt"))
    if ckpts:
        LATEST_CHECKPOINT = ckpts[-1]

if LATEST_CHECKPOINT and LATEST_CHECKPOINT.exists():
    print(f"\nResuming from {LATEST_CHECKPOINT}")
    ckpt = torch.load(LATEST_CHECKPOINT, map_location=device)
    # state_dict is from the raw model (saved without DataParallel wrapper)
    raw_model.load_state_dict(ckpt["model_state_dict"])
    optimizer.load_state_dict(ckpt["optimizer_state_dict"])
    scheduler.load_state_dict(ckpt["scheduler_state_dict"])
    global_step = ckpt["global_step"]
    best_val_loss = ckpt.get("best_val_loss", float('inf'))
    training_metrics = defaultdict(list, ckpt.get("metrics", {}))
    print(f"  Resumed at step {global_step}")
    del ckpt

print(f"\n--- Training Configuration ---")
print(f"  Batch size: {BATCH_SIZE}  (split across {torch.cuda.device_count()} GPU(s) by DataParallel)")
print(f"  Grad accum: {GRADIENT_ACCUMULATION_STEPS}")
print(f"  Effective batch: {BATCH_SIZE * GRADIENT_ACCUMULATION_STEPS}")
print(f"  Seq length: {SEQUENCE_LENGTH}")
print(f"  LR: {LEARNING_RATE}  |  Warmup: {WARMUP_STEPS}")
print(f"  Max steps: {MAX_TRAINING_STEPS}")
print(f"  Val every: {VALIDATION_INTERVAL}  |  Ckpt every: {CHECKPOINT_INTERVAL}")
print(f"  Precision: {MIXED_PRECISION}  |  Grad clip: {GRADIENT_CLIPPING}")
print(f"  Multi-GPU: {USE_MULTI_GPU_EFFECTIVE}  |  DataParallel: {IS_DATA_PARALLEL}")

print(f"\nStarting training from step {global_step} to {MAX_TRAINING_STEPS}...")
print("-" * 70)

train_iter = iter(train_loader)

while global_step < MAX_TRAINING_STEPS:
    optimizer.zero_grad()
    accum_loss = 0.0
    accum_steps_ok = 0

    for _ in range(GRADIENT_ACCUMULATION_STEPS):
        try:
            batch = next(train_iter)
        except StopIteration:
            train_iter = iter(train_loader)
            batch = next(train_iter)

        input_ids = batch["input_ids"].to(device)
        labels = batch["labels"].to(device)

        try:
            if autocast_dtype is not None:
                with torch.autocast(device_type=device.type, dtype=autocast_dtype):
                    outputs = model(input_ids=input_ids, labels=labels, return_dict=True)
                    loss = outputs.loss
            else:
                outputs = model(input_ids=input_ids, labels=labels, return_dict=True)
                loss = outputs.loss
        except RuntimeError as e:
            if "out of memory" in str(e):
                print(f"  CUDA OOM at step {global_step}!")
                OOM_OCCURRED = True
                torch.cuda.empty_cache()
                gc.collect()
                break
            raise

        if torch.isnan(loss) or torch.isinf(loss):
            print(f"  NaN/Inf loss at step {global_step} - skipping batch")
            del input_ids, labels, outputs, loss
            continue

        scaled = loss / GRADIENT_ACCUMULATION_STEPS
        if scaler: scaler.scale(scaled).backward()
        else: scaled.backward()

        accum_loss += loss.item()
        accum_steps_ok += 1
        del input_ids, labels, outputs, loss

    if accum_steps_ok == 0:
        continue

    if scaler: scaler.unscale_(optimizer)
    grad_norm = torch.nn.utils.clip_grad_norm_(model.parameters(), GRADIENT_CLIPPING)

    if scaler: scaler.step(optimizer); scaler.update()
    else: optimizer.step()
    scheduler.step()
    global_step += 1

    avg_loss = accum_loss / accum_steps_ok
    current_lr = optimizer.param_groups[0]['lr']

    training_metrics["step"].append(global_step)
    training_metrics["train_loss"].append(avg_loss)
    training_metrics["learning_rate"].append(current_lr)
    gn_val = grad_norm.item() if isinstance(grad_norm, torch.Tensor) else grad_norm
    training_metrics["grad_norm"].append(gn_val)

    if global_step % 10 == 0 or global_step <= 5:
        elapsed = time.time() - start_time
        sps = global_step / elapsed if elapsed > 0 else 0
        tokens = global_step * BATCH_SIZE * GRADIENT_ACCUMULATION_STEPS * SEQUENCE_LENGTH
        peak_vram = torch.cuda.max_memory_allocated()/(1024**3) if torch.cuda.is_available() else 0
        print(f"Step {global_step:>6d} | Loss:{avg_loss:.4f} | LR:{current_lr:.2e} | "
              f"Grad:{gn_val:.4f} | {sps:.2f} steps/s | {tokens:,} tokens | VRAM:{peak_vram:.1f}GB")

    # Validation
    if val_loader and global_step % VALIDATION_INTERVAL == 0:
        print(f"\n--- Validation at step {global_step} ---")
        raw_model.eval()
        val_losses = []
        with torch.no_grad():
            for vi, vbatch in enumerate(val_loader):
                vi_ids = vbatch["input_ids"].to(device)
                vi_lbl = vbatch["labels"].to(device)
                if autocast_dtype is not None:
                    with torch.autocast(device_type=device.type, dtype=autocast_dtype):
                        vo = model(input_ids=vi_ids, labels=vi_lbl, return_dict=True)
                        vl = vo.loss
                else:
                    vo = model(input_ids=vi_ids, labels=vi_lbl, return_dict=True)
                    vl = vo.loss
                val_losses.append(vl.item())
                del vi_ids, vi_lbl, vo, vl
                if vi >= 50: break

        val_loss = sum(val_losses) / len(val_losses) if val_losses else float('inf')
        ppl = math.exp(val_loss) if val_loss < 20 else float('inf')
        training_metrics["val_loss"].append(val_loss)
        training_metrics["val_step"].append(global_step)
        print(f"  Val loss: {val_loss:.4f} | PPL: {ppl:.2f}")

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            print(f"  New best val loss!")
            best_path = Path(CHECKPOINT_DIR) / "best_model.pt"
            torch.save({
                "model_state_dict": raw_model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "scheduler_state_dict": scheduler.state_dict(),
                "global_step": global_step, "best_val_loss": best_val_loss,
                "metrics": dict(training_metrics),
            }, best_path)
        raw_model.train()

    # Checkpoint
    if global_step % CHECKPOINT_INTERVAL == 0:
        ckpt_path = Path(CHECKPOINT_DIR) / f"training_checkpoint_step_{global_step}.pt"
        print(f"  Saving checkpoint {ckpt_path.name}...")
        torch.save({
            "model_state_dict": raw_model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "scheduler_state_dict": scheduler.state_dict(),
            "global_step": global_step, "best_val_loss": best_val_loss,
            "metrics": dict(training_metrics),
            "config": {"model_variant": MODEL_VARIANT, "batch_size": BATCH_SIZE,
                       "grad_accum": GRADIENT_ACCUMULATION_STEPS, "seq_length": SEQUENCE_LENGTH,
                       "learning_rate": LEARNING_RATE, "max_steps": MAX_TRAINING_STEPS,
                       "mixed_precision": MIXED_PRECISION, "seed": SEED,
                       "use_multi_gpu": USE_MULTI_GPU_EFFECTIVE},
        }, ckpt_path)
        meta = {"step": global_step, "train_loss": avg_loss, "best_val_loss": best_val_loss,
                "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")}
        with open(ckpt_path.with_suffix(".json"), "w") as f:
            json.dump(meta, f, indent=2)
        print(f"  Checkpoint saved")

    # Kaggle 9-hour safety check
    elapsed_hrs = (time.time() - start_time) / 3600
    if elapsed_hrs > 8.5:
        print(f"\n  WARNING: {elapsed_hrs:.2f}h elapsed — approaching Kaggle 9h limit.")
        print(f"  Saving emergency checkpoint and stopping.")
        emergency_path = Path(CHECKPOINT_DIR) / f"emergency_stop_step_{global_step}.pt"
        torch.save({
            "model_state_dict": raw_model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "scheduler_state_dict": scheduler.state_dict(),
            "global_step": global_step, "best_val_loss": best_val_loss,
            "metrics": dict(training_metrics),
        }, emergency_path)
        break

# Done
total_time = time.time() - start_time
final_train_loss = training_metrics["train_loss"][-1] if training_metrics["train_loss"] else float('inf')
total_tokens = global_step * BATCH_SIZE * GRADIENT_ACCUMULATION_STEPS * SEQUENCE_LENGTH

print(f"\n{'='*70}")
print(f"Training Complete!")
print(f"{'='*70}")
print(f"  Steps: {global_step:,}  |  Tokens: {total_tokens:,}")
print(f"  Time: {total_time/3600:.2f} hours")
print(f"  Final train loss: {final_train_loss:.4f}")
print(f"  Best val loss: {best_val_loss:.4f}")
if torch.cuda.is_available():
    print(f"  Peak VRAM (cuda:0): {torch.cuda.max_memory_allocated()/(1024**3):.2f} GB")
    print(f"  GPUs used: {torch.cuda.device_count()}")

TRAINING_COMPLETE = True
```

```python
import torch
from pathlib import Path

print("=" * 70)
print("Checkpoint Save/Load/Verify Test")
print("=" * 70)

test_ckpt = Path(CHECKPOINT_DIR) / "resume_test_checkpoint.pt"

print("\n1. Saving test checkpoint...")
raw_model.eval()
torch.save({
    "model_state_dict": raw_model.state_dict(),
    "optimizer_state_dict": optimizer.state_dict(),
    "scheduler_state_dict": scheduler.state_dict(),
    "global_step": global_step, "best_val_loss": best_val_loss,
    "metrics": dict(training_metrics),
}, test_ckpt)
print(f"  Size: {test_ckpt.stat().st_size/(1024**2):.1f} MB")

print("\n2. Reference output...")
test_in = torch.randint(0, raw_model.config.vocab_size, (1, min(SEQUENCE_LENGTH, 64)), device=device)
test_lb = torch.randint(0, raw_model.config.vocab_size, (1, min(SEQUENCE_LENGTH, 64)), device=device)
with torch.no_grad():
    ref = model(input_ids=test_in, labels=test_lb, return_dict=True)
    ref_loss = ref.loss.item()
    ref_logits = ref.logits[:, :5, :5].clone()
print(f"  Ref loss: {ref_loss:.6f}")

print("\n3. Reloading checkpoint...")
ckpt = torch.load(test_ckpt, map_location=device)
raw_model.load_state_dict(ckpt["model_state_dict"])
optimizer.load_state_dict(ckpt["optimizer_state_dict"])
scheduler.load_state_dict(ckpt["scheduler_state_dict"])
print(f"  Restored step: {ckpt['global_step']}")

print("\n4. Output after reload...")
raw_model.eval()
with torch.no_grad():
    new = model(input_ids=test_in, labels=test_lb, return_dict=True)
    new_loss = new.loss.item()
    new_logits = new.logits[:, :5, :5].clone()
print(f"  New loss: {new_loss:.6f}")

loss_diff = abs(ref_loss - new_loss)
logits_diff = (ref_logits - new_logits).abs().max().item()
print(f"\n5. Comparison:")
print(f"  Loss diff:   {loss_diff:.8f}")
print(f"  Logits diff: {logits_diff:.8f}")

RESUME_OK = loss_diff < 1e-5 and logits_diff < 1e-4
print(f"\n{'OK' if RESUME_OK else 'WARN'}: Checkpoint save/load {'verified' if RESUME_OK else 'discrepancy'}")

del test_in, test_lb, ref, new, ref_logits, new_logits, ckpt
raw_model.train()
print("=" * 70)
```

```python
import torch, json, hashlib, sys
from pathlib import Path
from datetime import datetime

print("=" * 70)
print("Release Artifact Export")
print("=" * 70)

release_dir = Path(FINAL_DIR) if FINAL_DIR else Path(KAGGLE_WORKING_DIR) / f"{MODEL_VARIANT}_release"
release_dir.mkdir(parents=True, exist_ok=True)
print(f"Release dir: {release_dir}")

# 1. Model checkpoint (saved from raw_model, no DP wrapper)
print("\n1. Final model checkpoint...")
final_ckpt = release_dir / "model.pt"
torch.save({
    "model_state_dict": raw_model.state_dict(),
    "global_step": global_step, "best_val_loss": best_val_loss,
    "xorzen_version": xorzen.__version__,
    "model_variant": MODEL_VARIANT,
}, final_ckpt)
print(f"  {final_ckpt.name}: {final_ckpt.stat().st_size/(1024**2):.1f} MB")

# 2. Config
print("\n2. Config...")
config_dict = {attr: getattr(raw_model.config, attr, None) for attr in
    ["model_name", "vocab_size", "hidden_size", "num_layers", "num_attention_heads",
     "context_length", "expert_count", "top_k_experts", "max_depth", "min_depth",
     "cot_dim", "cot_components", "tie_word_embeddings", "gradient_checkpointing"]}
config_dict["width_choices"] = list(raw_model.config.width_choices)
config_dict["model_variant"] = MODEL_VARIANT
config_dict["test_mode"] = USE_TEST_MODE
with open(release_dir / "config.json", "w") as f:
    json.dump(config_dict, f, indent=2)

# 3. Training config
print("\n3. Training config...")
train_cfg = {"batch_size": BATCH_SIZE, "gradient_accumulation_steps": GRADIENT_ACCUMULATION_STEPS,
             "sequence_length": SEQUENCE_LENGTH, "learning_rate": LEARNING_RATE,
             "weight_decay": WEIGHT_DECAY, "warmup_steps": WARMUP_STEPS,
             "max_training_steps": MAX_TRAINING_STEPS, "validation_interval": VALIDATION_INTERVAL,
             "checkpoint_interval": CHECKPOINT_INTERVAL, "gradient_clipping": GRADIENT_CLIPPING,
             "mixed_precision": MIXED_PRECISION, "seed": SEED, "dataset": DATASET_NAME,
             "tokenizer": TOKENIZER_NAME, "use_multi_gpu": USE_MULTI_GPU_EFFECTIVE,
             "num_gpus": torch.cuda.device_count() if torch.cuda.is_available() else 0,
             "max_dataset_samples": MAX_DATASET_SAMPLES}
with open(release_dir / "training_config.json", "w") as f:
    json.dump(train_cfg, f, indent=2)

# 4. Training metrics
print("\n4. Training metrics...")
with open(release_dir / "training_metrics.json", "w") as f:
    json.dump(dict(training_metrics), f, indent=2)

# 5. Validation metrics
print("\n5. Validation metrics...")
if 'val_metrics' in dir():
    with open(release_dir / "validation_metrics.json", "w") as f:
        json.dump(val_metrics, f, indent=2)

# 6. Generation samples
print("\n6. Generation samples...")
with open(release_dir / "generation_samples.json", "w") as f:
    json.dump(generation_samples, f, indent=2)

# 7. Environment
print("\n7. Environment info...")
env = {"python_version": f"{sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}",
       "pytorch_version": torch.__version__,
       "cuda_version": torch.version.cuda if torch.cuda.is_available() else None,
       "gpu_name": torch.cuda.get_device_name(0) if torch.cuda.is_available() else None,
       "gpu_count": torch.cuda.device_count() if torch.cuda.is_available() else 0,
       "xorzen_version": xorzen.__version__,
       "platform": "Kaggle"}
with open(release_dir / "environment.txt", "w") as f:
    for k, v in env.items(): f.write(f"{k}: {v}\n")

# 8. SHA-256 hashes
print("\n8. Checksums...")
hashes = {}
for fname in ["model.pt", "config.json", "training_config.json", "training_metrics.json"]:
    fp = release_dir / fname
    if fp.exists():
        h = hashlib.sha256(fp.read_bytes()).hexdigest()
        hashes[fname] = h
        print(f"  {fname}: {h[:16]}...")
with open(release_dir / "checksums.json", "w") as f:
    json.dump(hashes, f, indent=2)

print(f"\nOK: Release exported to {release_dir}")
print(f"     Downloadable from Kaggle UI: /kaggle/working/{release_dir.relative_to(KAGGLE_WORKING_DIR)}")
print("=" * 70)
```

## 17. Training Curves

Plot loss / LR / grad-norm / val-loss as PNG, saved to `/kaggle/working/`.

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path

print("Generating training curves...")

steps = training_metrics.get("step", [])
losses = training_metrics.get("train_loss", [])
lrs = training_metrics.get("learning_rate", [])
gns = training_metrics.get("grad_norm", [])
val_steps = training_metrics.get("val_step", [])
val_losses = training_metrics.get("val_loss", [])

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle(f"XORZEN {MODEL_VARIANT} Training Metrics (Kaggle T4)", fontsize=14, fontweight='bold')

if steps and losses:
    axes[0,0].plot(steps, losses, color='#2563eb', linewidth=0.8, alpha=0.8)
    axes[0,0].set_title("Training Loss"); axes[0,0].set_xlabel("Step"); axes[0,0].set_ylabel("Loss")
    axes[0,0].grid(True, alpha=0.3)

if val_steps and val_losses:
    axes[0,1].plot(val_steps, val_losses, color='#dc2626', marker='o', markersize=4, linewidth=1)
    axes[0,1].set_title("Validation Loss"); axes[0,1].set_xlabel("Step"); axes[0,1].set_ylabel("Val Loss")
    axes[0,1].grid(True, alpha=0.3)
else:
    axes[0,1].text(0.5, 0.5, "No validation data", ha='center', va='center', transform=axes[0,1].transAxes)
    axes[0,1].set_title("Validation Loss (no data)")

if steps and lrs:
    axes[1,0].plot(steps, lrs, color='#16a34a', linewidth=1)
    axes[1,0].set_title("Learning Rate"); axes[1,0].set_xlabel("Step"); axes[1,0].set_ylabel("LR")
    axes[1,0].grid(True, alpha=0.3)

if steps and gns:
    gn_clip = np.clip(gns, 0, np.percentile(gns, 99) if len(gns) > 10 else max(gns))
    axes[1,1].plot(steps, gn_clip, color='#9333ea', linewidth=0.5, alpha=0.7)
    axes[1,1].set_title("Gradient Norm"); axes[1,1].set_xlabel("Step"); axes[1,1].set_ylabel("Grad Norm")
    axes[1,1].grid(True, alpha=0.3)

plt.tight_layout()
curves_path = Path(FINAL_DIR) / "training_curves.png" if FINAL_DIR else Path(KAGGLE_WORKING_DIR) / "training_curves.png"
plt.savefig(curves_path, dpi=150, bbox_inches='tight')
plt.show()
print(f"Saved to {curves_path}")
```
